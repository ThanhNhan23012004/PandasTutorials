```python
s1 = pd.Series([i for i in range(20)])

print(s1.min())
print(s1.max())

# Print the 25th percentile
print(s1.quantile(0.25))

# Print the 75th percentile
print(s1.quantile(0.75))

# Print the median
print(s1.median())
```

```python
import numpy as np
np.percentile(s1, q=[0, 25, 50, 75, 100])
```
